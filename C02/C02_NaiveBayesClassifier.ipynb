```python
def build_histogram(train_col, num_bins, plot=False, title=""):

    min_val = train_col.min()
    max_val = train_col.max()

    bin_edges = np.linspace(min_val, max_val, num_bins + 1)
    bin_edges[0]  = -np.inf
    bin_edges[-1] =  np.inf

    bin_indices = np.digitize(train_col.dropna(), bin_edges) - 1
    hist = np.bincount(bin_indices, minlength=num_bins)

    if plot:
        plt.fill_between(
            bin_edges.repeat(2)[1:-1],
            hist.repeat(2),
            facecolor='steelblue'
        )
        plt.title(title)
        plt.xlabel("Value")
        plt.ylabel("Count")
        plt.show()

    return hist, bin_edges
```

```python
def build_likelihood(col_train, num_bins=40):
    # drop missing values (NaN or -1)
    col_train = col_train[~col_train.isna()]
    col_train = col_train[col_train != -1]

    # compute bin edges
    min_val = col_train.min()
    max_val = col_train.max()

    bin_edges = np.linspace(min_val, max_val, num_bins + 1)
    bin_edges[0]  = -np.inf
    bin_edges[-1] =  np.inf

    # assign to bins
    bin_idx = np.digitize(col_train, bin_edges) - 1

    # histogram
    hist = np.bincount(bin_idx, minlength=num_bins)

    # MLE: normalize → categorical likelihood
    prob = hist / hist.sum()

    return prob, bin_edges
```

```python
def plot_likelihoods_for_feature(X_train, y_train, col, num_bins=40, title=None):
    plt.figure(figsize=(7, 4))

    for cls, color in zip([0, 1], ["steelblue", "darkorange"]):
        # select class-conditional data
        col_train = X_train.loc[y_train == cls, col]

        prob, bin_edges = build_likelihood(col_train, num_bins)

        # step-style x and y
        x = bin_edges.repeat(2)[1:-1]
        y = prob.repeat(2)

        plt.plot(x, y, color=color, label=f"Attrition = {cls}")

    plt.xlabel(col)
    plt.ylabel("Likelihood  P(x | y)")
    plt.title(title if title else f"{col} likelihoods by Attrition")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
def build_histogram_likelihood(col_train, num_bins=10, alpha=1.0):
    # drop missing values
    col_train = col_train.dropna()

    # bin edges
    min_val = col_train.min()
    max_val = col_train.max()

    bin_edges = np.linspace(min_val, max_val, num_bins + 1)
    bin_edges[0]  = -np.inf
    bin_edges[-1] =  np.inf

    # assign bins
    bin_idx = np.digitize(col_train, bin_edges) - 1

    # counts
    hist = np.bincount(bin_idx, minlength=num_bins)

    # Laplace smoothing (add-α)
    hist = hist + alpha

    # normalize → likelihood
    prob = hist / hist.sum()

    return prob, bin_edges
```

```python
def build_categorical_likelihood(col_train, alpha=1.0):
    # drop missing-category markers
    col_train = col_train.dropna()
    col_train = col_train[col_train != -1]

    # counts
    hist = np.bincount(col_train.astype(int))

    # Laplace smoothing
    hist = hist + alpha

    # normalize
    prob = hist / hist.sum()

    return prob
```

```python
def train_naive_bayes(X_train, y_train,
                      continuous_cols,
                      categorical_cols,
                      num_bins=10,
                      alpha=1.0):

    model = {}

    # ----- prior -----
    N = len(y_train)
    model["prior"] = {
        0: np.mean(y_train == 0),
        1: np.mean(y_train == 1)
    }

    # ----- likelihoods -----
    model["likelihoods"] = {0: {}, 1: {}}
    model["bin_edges"]  = {0: {}, 1: {}}

    for cls in [0, 1]:
        X_cls = X_train.loc[y_train == cls]

        # continuous features
        for col in continuous_cols:
            prob, edges = build_histogram_likelihood(
                X_cls[col],
                num_bins=num_bins,
                alpha=alpha
            )

            model["likelihoods"][cls][col] = prob
            model["bin_edges"][cls][col]   = edges

        # categorical features
        for col in categorical_cols:
            prob = build_categorical_likelihood(
                X_cls[col],
                alpha=alpha
            )

            model["likelihoods"][cls][col] = prob

    return model
```

```python
def predict_one(x_row, model, continuous_cols, categorical_cols):
    log_post = {}

    for cls in [0, 1]:
        # start with log prior
        log_p = np.log(model["prior"][cls])

        # continuous features
        for col in continuous_cols:
            x = x_row[col]

            if pd.isna(x):
                continue  # skip missing

            edges = model["bin_edges"][cls][col]
            prob  = model["likelihoods"][cls][col]

            b = np.digitize(x, edges) - 1
            b = np.clip(b, 0, len(prob) - 1)

            log_p += np.log(prob[b])

        # categorical features
        for col in categorical_cols:
            x = x_row[col]

            if x == -1:
                continue  # skip missing category

            prob = model["likelihoods"][cls][col]

            if x < len(prob):
                log_p += np.log(prob[int(x)])
            else:
                # unseen category → tiny probability
                log_p += np.log(1e-12)

        log_post[cls] = log_p

    # choose class with max posterior
    return max(log_post, key=log_post.get)
```

```python
def predict_naive_bayes(X_test, model, continuous_cols, categorical_cols):
    y_pred = []

    for _, row in X_test.iterrows():
        y_hat = predict_one(row, model, continuous_cols, categorical_cols)
        y_pred.append(y_hat)

    return np.array(y_pred)
```

```python
def classification_metrics(y_true, y_pred):
    TP = np.sum((y_true == 1) & (y_pred == 1))
    TN = np.sum((y_true == 0) & (y_pred == 0))
    FP = np.sum((y_true == 0) & (y_pred == 1))
    FN = np.sum((y_true == 1) & (y_pred == 0))

    accuracy  = (TP + TN) / (TP + TN + FP + FN)
    precision = TP / (TP + FP + 1e-12)
    recall    = TP / (TP + FN + 1e-12)
    f1        = 2 * precision * recall / (precision + recall + 1e-12)

    return accuracy, precision, recall, f1
```

```python
from scipy.stats import norm

# ---------- Metrics ----------
def classification_metrics(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    TP = np.sum((y_true == 1) & (y_pred == 1))
    TN = np.sum((y_true == 0) & (y_pred == 0))
    FP = np.sum((y_true == 0) & (y_pred == 1))
    FN = np.sum((y_true == 1) & (y_pred == 0))

    accuracy  = (TP + TN) / (TP + TN + FP + FN + 1e-12)
    precision = TP / (TP + FP + 1e-12)
    recall    = TP / (TP + FN + 1e-12)
    f1        = 2 * precision * recall / (precision + recall + 1e-12)

    return accuracy, precision, recall, f1


# ---------- Categorical likelihood (with Laplace smoothing) ----------
def build_categorical_likelihood(col_train, alpha=1.0):
    # drop NaN first
    col_train = col_train.dropna()

    # drop missing-category marker
    col_train = col_train[col_train != -1]

    if len(col_train) == 0:
        return np.array([1.0])  # degenerate fallback

    hist = np.bincount(col_train.astype(int))
    hist = hist + alpha
    prob = hist / hist.sum()

    return prob


# ---------- Train: Gaussian for continuous, categorical counts for categorical ----------
def train_gaussian_nb(X_train, y_train, continuous_cols, categorical_cols, alpha=1.0):
    model = {
        "prior": {},
        "gauss": {0: {}, 1: {}},      # mean/std per feature per class
        "cat":   {0: {}, 1: {}}       # prob array per feature per class
    }

    # priors
    model["prior"][0] = np.mean(y_train == 0)
    model["prior"][1] = np.mean(y_train == 1)

    for cls in [0, 1]:
        X_cls = X_train.loc[y_train == cls]

        # continuous: estimate mean/std (MLE-ish)
        for col in continuous_cols:
            s = X_cls[col].dropna()
            if len(s) == 0:
                # if all missing for that class/feature, store None
                model["gauss"][cls][col] = (None, None)
                continue

            mean = float(s.mean())
            std = float(s.std(ddof=0))  # MLE variance uses ddof=0

            # avoid std=0 -> pdf problems
            if std < 1e-9:
                std = 1e-9

            model["gauss"][cls][col] = (mean, std)

        # categorical: estimate P(x=k|y) with Laplace smoothing
        for col in categorical_cols:
            prob = build_categorical_likelihood(X_cls[col], alpha=alpha)
            model["cat"][cls][col] = prob

    return model


# ---------- Predict one row ----------
def predict_one_gaussian_nb(row, model, continuous_cols, categorical_cols):
    log_post = {}

    for cls in [0, 1]:
        log_p = np.log(model["prior"][cls] + 1e-12)

        # continuous likelihoods via Gaussian pdf
        for col in continuous_cols:
            x = row[col]
            if pd.isna(x):
                continue  # missing -> drop term

            mean, std = model["gauss"][cls][col]
            if mean is None:
                continue  # no info for this feature in this class

            # log pdf
            log_p += np.log(norm(mean, std).pdf(float(x)) + 1e-300)

        # categorical likelihoods via counts
        for col in categorical_cols:
            x = row[col]
            if x == -1 or pd.isna(x):
                continue  # missing -> drop term

            prob = model["cat"][cls][col]
            xi = int(x)

            # unseen category code -> give tiny prob (or could smooth over max)
            if xi < 0 or xi >= len(prob):
                log_p += np.log(1e-12)
            else:
                log_p += np.log(prob[xi] + 1e-300)

        log_post[cls] = log_p

    return 1 if log_post[1] > log_post[0] else 0


# ---------- Predict batch ----------
def predict_gaussian_nb(X_test, model, continuous_cols, categorical_cols):
    preds = []
    for _, row in X_test.iterrows():
        preds.append(predict_one_gaussian_nb(row, model, continuous_cols, categorical_cols))
    return np.array(preds)


# =========================
# T12: Train -> Predict -> Report
# =========================
# Use your feature split:
continuous_cols = [
    "Age", "DailyRate", "DistanceFromHome", "HourlyRate", "MonthlyIncome", "MonthlyRate"
]

categorical_cols = [
    "BusinessTravel", "Department", "EducationField", "Education", "EnvironmentSatisfaction", "Gender", "JobInvolvement", "JobLevel",
    "JobRole", "JobSatisfaction", "MaritalStatus", "NumCompaniesWorked", "OverTime", "PercentSalaryHike", "RelationshipSatisfaction",
    "StockOptionLevel", "TotalWorkingYears", "TrainingTimesLastYear", "WorkLifeBalance", "YearsAtCompany", "YearsInCurrentRole",
    "YearsSinceLastPromotion", "YearsWithCurrManager"
]

model_gnb = train_gaussian_nb(
    X_train, y_train,
    continuous_cols=continuous_cols,
    categorical_cols=categorical_cols,
    alpha=1.0
)

y_pred = predict_gaussian_nb(X_test, model_gnb, continuous_cols, categorical_cols)

acc, prec, rec, f1 = classification_metrics(y_test.values, y_pred)
print("Accuracy :", acc)
print("Precision:", prec)
print("Recall   :", rec)
print("F1 score :", f1)
```

```python
def log_posterior_diff(row, model, continuous_cols, categorical_cols):
    log_p = {}

    for cls in [0, 1]:
        lp = np.log(model["prior"][cls] + 1e-12)

        # continuous features
        for col in continuous_cols:
            x = row[col]
            if pd.isna(x):
                continue
            mean, std = model["gauss"][cls][col]
            if mean is None:
                continue
            lp += np.log(norm(mean, std).pdf(float(x)) + 1e-300)

        # categorical features
        for col in categorical_cols:
            x = row[col]
            if pd.isna(x) or x == -1:
                continue
            prob = model["cat"][cls][col]
            xi = int(x)
            if xi < len(prob):
                lp += np.log(prob[xi] + 1e-300)
            else:
                lp += np.log(1e-12)

        log_p[cls] = lp

    return log_p[1] - log_p[0]
```

```python

def roc_curve_from_scores(y_true, scores, thresholds):
    tprs, fprs = [], []
    y_true = np.asarray(y_true)

    for t in thresholds:
        y_pred = (scores > t).astype(int)

        TP = np.sum((y_true == 1) & (y_pred == 1))
        TN = np.sum((y_true == 0) & (y_pred == 0))
        FP = np.sum((y_true == 0) & (y_pred == 1))
        FN = np.sum((y_true == 1) & (y_pred == 0))

        tpr = TP / (TP + FN + 1e-12)   # recall
        fpr = FP / (FP + TN + 1e-12)

        tprs.append(tpr)
        fprs.append(fpr)

    return np.array(fprs), np.array(tprs)
```

```python
def score_gaussian_nb(row, model, continuous_cols, categorical_cols):
    log_p = {}
    for cls in [0, 1]:
        lp = np.log(model["prior"][cls] + 1e-12)

        # continuous: Gaussian pdf
        for col in continuous_cols:
            x = row[col]
            if pd.isna(x):
                continue
            mean, std = model["gauss"][cls][col]
            if mean is None:
                continue
            lp += np.log(norm(mean, std).pdf(float(x)) + 1e-300)

        # categorical: categorical likelihood
        for col in categorical_cols:
            x = row[col]
            if pd.isna(x) or x == -1:
                continue
            prob = model["cat"][cls][col]
            xi = int(x)
            if 0 <= xi < len(prob):
                lp += np.log(prob[xi] + 1e-300)
            else:
                lp += np.log(1e-12)

        log_p[cls] = lp

    return log_p[1] - log_p[0]
```

```python
def score_hist_nb(row, model, continuous_cols, categorical_cols):
    log_p = {}
    for cls in [0, 1]:
        lp = np.log(model["prior"][cls] + 1e-12)

        # continuous: histogram bins
        for col in continuous_cols:
            x = row[col]
            if pd.isna(x):
                continue
            edges = model["bin_edges"][cls][col]
            probs = model["likelihoods"][cls][col]
            b = np.digitize(x, edges) - 1
            b = np.clip(b, 0, len(probs) - 1)
            lp += np.log(probs[b] + 1e-300)

        # categorical: categorical likelihood
        for col in categorical_cols:
            x = row[col]
            if pd.isna(x) or x == -1:
                continue
            probs = model["likelihoods"][cls][col]
            xi = int(x)
            if 0 <= xi < len(probs):
                lp += np.log(probs[xi] + 1e-300)
            else:
                lp += np.log(1e-12)

        log_p[cls] = lp

    return log_p[1] - log_p[0]
```
