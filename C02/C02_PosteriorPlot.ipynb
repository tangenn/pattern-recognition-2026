```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- 1. Gaussian PDF function ----------
def gaussian_pdf(x, mu, var):
    return 1 / np.sqrt(2 * np.pi * var) * np.exp(-(x - mu)**2 / (2 * var))

# ---------- 2. Parameters from the problem ----------
mu1, var1 = 4, 2   # class w1: N(4, 2)
mu2, var2 = 0, 2   # class w2: N(0, 2)

# ---------- 3. Range of x values ----------
x = np.linspace(-6, 10, 1000)

# ---------- 4. Compute likelihoods ----------
p1 = gaussian_pdf(x, mu1, var1)   # p(x | w1)
p2 = gaussian_pdf(x, mu2, var2)   # p(x | w2)

# ---------- 5. Compute posteriors (equal priors) ----------
post1 = p1 / (p1 + p2)   # P(w1 | x)
post2 = p2 / (p1 + p2)   # P(w2 | x)

# ---------- 6. Decision boundary (likelihood ratio = 1) ----------
# Since variances are equal, boundary is midpoint
x_star = (mu1 + mu2) / 2   # = 2

# ---------- 7. Plot ----------
plt.figure(figsize=(8, 5))

plt.plot(x, post1, label="P(w1 | x)")
plt.plot(x, post2, label="P(w2 | x)")

# draw decision boundary
plt.axvline(x_star, linestyle="--", label=f"Decision boundary x = {x_star}")

plt.xlabel("x (amount of food eaten)")
plt.ylabel("Posterior probability")
plt.title("Posterior probabilities for the two classes")
plt.ylim(0, 1)
plt.legend()

plt.show()
```
