```python
import numpy as np
```

```python
X = np.array(
    train[["Pclass", "Sex", "Age", "Embarked"]].values,
    dtype=float
)

y = np.array(train["Survived"].values, dtype=float)
```

```python
X.shape, y.shape
```

```python
m = X.shape[0]
X = np.c_[np.ones(m), X]
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
def predict_prob(X, w):
    return sigmoid(X @ w)
def compute_loss(X, y, w):
    m = len(y)
    y_hat = predict_prob(X, w)
    epsilon = 1e-8  # numerical stability
    return -(1/m) * np.sum(
        y * np.log(y_hat + epsilon) +
        (1 - y) * np.log(1 - y_hat + epsilon)
    )
def compute_gradient(X, y, w):
    m = len(y)
    y_hat = predict_prob(X, w)
    return (1/m) * (X.T @ (y_hat - y))
def gradient_descent(X, y, lr=0.1, decay=1e-3, epochs=5000):
    w = np.zeros(X.shape[1])
    losses = []

    for i in range(epochs):
        lr_t = lr / (1 + decay * i)  

        grad = compute_gradient(X, y, w)
        w -= lr_t * grad

        if i % 500 == 0:
            loss = compute_loss(X, y, w)
            losses.append(loss)
            print(f"Epoch {i}: lr={lr_t:.5f}, loss={loss:.4f}")

    return w, losses
```

```python
w, losses = gradient_descent(X, y, epochs=1000000)
```

```python
y_train_prob = predict_prob(X, w)
```

```python
y_train_pred = (y_train_prob >= 0.5).astype(int)
```

```python
accuracy = (y_train_pred == y).mean()
accuracy
```

```python
X_base = X
```

```python
def add_bias(X):
    return np.c_[np.ones(X.shape[0]), X]

def accuracy_from_probs(y_prob, y_true, thresh=0.5):
    y_pred = (y_prob >= thresh).astype(int)
    return (y_pred == y_true).mean()
```

```python
Pclass = X_base[:, 0]
Sex   = X_base[:, 1]
Age   = X_base[:, 2]
Emb   = X_base[:, 3]

X_poly = np.column_stack([
    X_base,
    Age**2,          # x^2 example
    Sex*Age,         # x1*x2 example
    Pclass*Age,
    Pclass*Sex,
    Age*Emb
])
```

```python
X_poly_b = add_bias(X_poly)
w_poly, losses_poly = gradient_descent(X_poly_b, y, lr=0.01, epochs=5000)
```

```python
y_prob_poly = predict_prob(X_poly_b, w_poly)
train_acc_poly = accuracy_from_probs(y_prob_poly, y)
train_acc_poly
```

```python
def build_poly_features(df):
    X_base = df[["Pclass", "Sex", "Age", "Embarked"]].to_numpy(dtype=float)

    Pclass = X_base[:, 0]
    Sex    = X_base[:, 1]
    Age    = X_base[:, 2]
    Emb    = X_base[:, 3]

    X_poly = np.column_stack([
        X_base,
        Age**2,
        Sex*Age,
        Pclass*Age,
        Pclass*Sex,
        Age*Emb
    ])
    return X_poly
```

```python
X_train_poly = build_poly_features(train)
y_train = train["Survived"].to_numpy(dtype=float)

X_train_poly_b = add_bias(X_train_poly)

w_poly, losses_poly = gradient_descent(X_train_poly_b, y_train, lr=0.01, epochs=5000)
print("Train poly shape:", X_train_poly_b.shape, "w_poly shape:", w_poly.shape)
```

```python
X_test_poly = build_poly_features(test)
X_test_poly_b = add_bias(X_test_poly)

print("Test poly shape:", X_test_poly_b.shape, "w_poly shape:", w_poly.shape)

y_test_prob = predict_prob(X_test_poly_b, w_poly)
y_test_pred = (y_test_prob >= 0.5).astype(int)
```

```python
import pandas as pd

submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": y_test_pred
})

submission.to_csv("submission_poly.csv", index=False)
```

```python
X_lin = train[["Pclass", "Sex", "Age", "Embarked"]].to_numpy(dtype=float)
X_lin_b = add_bias(X_lin)
y = train["Survived"].to_numpy(dtype=float)
```

```python
def linear_predict(X, w):
    return X @ w

def linear_mse(X, y, w):
    e = linear_predict(X, w) - y
    return np.mean(e**2)

def linear_gd(X, y, lr=0.01, epochs=5000):
    w = np.zeros(X.shape[1])
    for i in range(epochs):
        grad = (2/len(y)) * (X.T @ (X @ w - y))  # gradient of MSE
        w -= lr * grad
        if i % 500 == 0:
            print(f"Epoch {i}: MSE={linear_mse(X,y,w):.4f}")
    return w
```

```python
w_lin_gd = linear_gd(X_lin_b, y, lr=0.01, epochs=5000)
```

```python
y_hat = linear_predict(X_lin_b, w_lin_gd)
y_pred = (y_hat >= 0.5).astype(int)
train_acc_lin = (y_pred == y).mean()
train_acc_lin
```

```python
X_test_lin = test[["Pclass", "Sex", "Age", "Embarked"]].to_numpy(dtype=float)
X_test_lin_b = add_bias(X_test_lin)

y_hat_test = linear_predict(X_test_lin_b, w_lin_gd)
y_test_pred = (y_hat_test >= 0.5).astype(int)

submission = pd.DataFrame({"PassengerId": test["PassengerId"], "Survived": y_test_pred})
submission.to_csv("submission_linear_gd.csv", index=False)
```
